```python
from collections import defaultdict
import numpy as np
```

```python
d = defaultdict(int)
c = 0
d_indexes = {}
```

```python
with open("animal_farm.txt") as file:
    for line in file.readlines():
        for word in line.split():
            d[word] += 1

            if d_indexes.get(word) is None:
                d_indexes[word] = c
                c += 1

```

```python
def vectorize(word: str):
    vec = np.zeros(len(d))
    index = d_indexes.get(word)

    if index is None:
        return None

    vec[index] = 1
    return vec
```
